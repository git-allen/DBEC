# E-Commerce Product Categorization (Retail)

Embedding product descriptions to find products that are identical in description but split across store departments (e.g., 'Smart Thermostats' split between Home Goods and Electronics).

```python
import sys
sys.path.append('..')
import numpy as np
from dbec import DBEC
```

### 1. Simulate Data
We generate dummy embeddings and assign conflicting labels to simulate an 'impure' region.

```python
# Generate 100 random embeddings of dimension 128
embeddings = np.random.rand(100, 128)

# Assign specific domain labels
labels = np.array(['Home Goods', 'Electronics', 'Electronics', 'Home Goods', 'Hardware'] * 20)
```

### 2. Run DBEC
We run the Density-Based Entropy Clustering algorithm to find the regions where labels are highly diverse despite spatial proximity.

```python
model = DBEC(neighbor=5, min_points_entropy=5, min_samples_high=0.3, entropy_threshold=0.3)
impure_clusters = model.fit_predict(embeddings, labels)
print("Cluster Assignments:", impure_clusters[:10], "...")
```

### 3. Extract Analytics
We pull the DBEC run metrics to understand the composition of the impure clusters found.

```python
metrics = model.get_run_metrics(embeddings, labels)
print("Number of Impure Clusters Found:", metrics['num_clusters'])
print("Total Core Points in Clusters:", metrics['number_core'])
print("Max Entropy of Dataset:", round(metrics['max_entropy'], 2))
```

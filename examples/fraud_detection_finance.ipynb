# Fraud Detection & System Auditing (Finance)

In this example, we embed financial transaction metadata. We use DBEC to find dense clusters of identical transactions where some are tagged 'Legitimate' and others 'Fraudulent', indicating a new fraud tactic.

```python
import sys
sys.path.append('..')
import numpy as np
from dbec import DBEC
```

### 1. Simulate Data
We generate dummy embeddings and assign conflicting labels to simulate an 'impure' region.

```python
# Generate 100 random embeddings of dimension 10
embeddings = np.random.rand(100, 10)

# Assign specific domain labels
labels = np.array(['Legitimate', 'Legitimate', 'Legitimate', 'Fraudulent', 'Fraudulent'] * 20)
```

### 2. Run DBEC
We run the Density-Based Entropy Clustering algorithm to find the regions where labels are highly diverse despite spatial proximity.

```python
model = DBEC(neighbor=5, min_points_entropy=5, min_samples_high=0.3, entropy_threshold=0.3)
impure_clusters = model.fit_predict(embeddings, labels)
print("Cluster Assignments:", impure_clusters[:10], "...")
```

### 3. Extract Analytics
We pull the DBEC run metrics to understand the composition of the impure clusters found.

```python
from collections import Counter
metrics = model.get_run_metrics(embeddings, labels)
print("Number of Impure Clusters Found:", metrics['num_clusters'])
if metrics['num_clusters'] > 0:
    cluster_0_labels = [labels[i] for i, clusters in enumerate(impure_clusters) if 0 in clusters]
    print("Cluster 0 Composition:", dict(Counter(cluster_0_labels)))
print("Total Core Points in Clusters:", metrics['number_core'])
```
